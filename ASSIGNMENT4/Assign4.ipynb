```python
import pandas as pd

fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]
entry1 = {
    "question": "how do i update my registered mobile number",
    "answer": "Go to Settings and update your registered mobile number.",
    "keywords": "mobile number update account",
    "category": "account"
}
entry2 = {
    "question": "where can i get general help",
    "answer": "You can contact the help desk for general assistance.",
    "keywords": "help support assistance",
    "category": "general"
}

entries = fixed_entries + [entry1, entry2]
df = pd.DataFrame(entries)

print(df)
```

```python
def score_query(query, df):
    query_words = query.lower().split()

    results = []

    for index, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()

        score = 0

        for word in query_words:
            if word in text:
                score += 1

        if score > 0:
            results.append((score, row["question"]))

    results.sort(reverse=True)
    return results
query = input("Enter your question: ")
results = score_query(query, df)
print("Matching Entries:")
for score, question in results:
    print("Score:", score, "|", question)
```

```python
def same_category(category_name, df):
    return df[df["category"] == category_name]
category_name = entry1["category"]
print("Category:", category_name)
print(same_category(category_name, df))
```

```python
def best_matches(query, df):
    query_words = query.lower().split()
    scores = []

    for index, row in df.iterrows():
        text = row["question"].lower() + " " + row["keywords"].lower()
        score = 0

        for word in query_words:
            if word in text:
                score += 1

        scores.append(score)
    highest_score = max(scores)
    print("Highest Score:", highest_score)
    print("Matching Entries:")

    for i in range(len(df)):
        if scores[i] == highest_score:
            print(df.loc[i, "question"])
print("Tie Query:")
best_matches("fee", df)
print("\nNon-Tie Query:")
best_matches("password", df)
```
