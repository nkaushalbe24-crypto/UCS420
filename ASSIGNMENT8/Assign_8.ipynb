```python

import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from collections import Counter

# Download required NLTK resources
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

# Unique paragraph about technology
paragraph = """
Technology makes learning easier and more flexible.
Students use digital tools to explore ideas and solve problems.
Artificial intelligence can explain difficult topics in simple language.
Online resources help learners practise skills at their own pace.
However, students should use technology responsibly and think critically.
"""

# 1. Convert text to lowercase and remove punctuation
clean_text = re.sub(r"[^a-zA-Z0-9\s]", "", paragraph.lower())
print("Cleaned text:\n", clean_text)

# 2. Tokenize into words and sentences
words = word_tokenize(paragraph.lower())
sentences = sent_tokenize(paragraph)

print("\nWord tokens:", words)
print("\nSentence tokens:", sentences)

# 3. Compare split() and word_tokenize()
split_words = clean_text.split()
nltk_words = word_tokenize(paragraph.lower())

print("\nUsing split():", split_words)
print("\nUsing word_tokenize():", nltk_words)

# 4. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [
    word for word in word_tokenize(clean_text)
    if word not in stop_words
]

print("\nWords after stopword removal:", filtered_words)

# 5. Word frequency distribution
frequency = Counter(filtered_words)

print("\nWord frequency distribution:")
for word, count in frequency.most_common():
    print(word, ":", count)
```

```python

import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download required resources
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

paragraph = """
Technology makes learning easier and more flexible.
Students use digital tools to explore ideas and solve problems.
Artificial intelligence can explain difficult topics in simple language.
Online resources help learners practise skills at their own pace.
However, students should use technology responsibly and think critically.
"""

# 1. Extract words containing only alphabets
words = re.findall(r"\b[a-zA-Z]+\b", paragraph.lower())
print("Alphabet-only words:", words)

# 2. Remove stopwords
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]
print("\nAfter stopword removal:", filtered_words)

# 3. Stemming using PorterStemmer
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
print("\nStemmed words:", stemmed_words)

# 4. Lemmatization using WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_words
]
print("\nLemmatized words:", lemmatized_words)

# 5. Compare results
print("\nOriginal word -> Stemmed word -> Lemmatized word")
for word, stem, lemma in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    print(word, "->", stem, "->", lemma)
```

```python

import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Three short texts
texts = [
    "The camera takes sharp photos and records clear video.",
    "The phone has a bright screen and a powerful battery.",
    "The camera phone captures sharp photos with a bright screen."
]

# 1. Bag of Words representation
count_vectorizer = CountVectorizer(stop_words='english')
bow_matrix = count_vectorizer.fit_transform(texts)

print("Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words matrix:")
print(bow_matrix.toarray())

# 2. TF-IDF representation
tfidf_vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

terms = tfidf_vectorizer.get_feature_names_out()

print("\nTF-IDF matrix:")
print(np.round(tfidf_matrix.toarray(), 3))

# 3. Top 3 keywords from each text
for i in range(len(texts)):
    scores = tfidf_matrix.toarray()[i]
    top_indices = scores.argsort()[::-1][:3]

    print("\nTop 3 keywords in Text", i + 1)

    for index in top_indices:
        if scores[index] > 0:
            print(terms[index], ":", round(scores[index], 3))
```

```python

import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Two texts about different technologies
text_ai = """
Artificial intelligence learns patterns from data.
AI systems can predict outcomes and automate tasks.
"""

text_blockchain = """
Blockchain stores records in linked digital blocks.
A distributed network verifies transactions.
"""

# 1. Preprocess and tokenize
def preprocess(text):
    text = re.sub(r"[^a-zA-Z\s]", "", text.lower())
    return text.split()

tokens_ai = preprocess(text_ai)
tokens_blockchain = preprocess(text_blockchain)

print("AI tokens:", tokens_ai)
print("\nBlockchain tokens:", tokens_blockchain)

# 2(a). Jaccard similarity
set_ai = set(tokens_ai)
set_blockchain = set(tokens_blockchain)

intersection = set_ai.intersection(set_blockchain)
union = set_ai.union(set_blockchain)

jaccard = len(intersection) / len(union)

print("\nCommon words:", intersection)
print("Jaccard similarity:", round(jaccard, 4))

# 2(b). Cosine similarity using TF-IDF
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(
    [text_ai, text_blockchain]
)

cosine = cosine_similarity(
    tfidf_matrix[0:1],
    tfidf_matrix[1:2]
)[0][0]

print("Cosine similarity:", round(cosine, 4))
```
