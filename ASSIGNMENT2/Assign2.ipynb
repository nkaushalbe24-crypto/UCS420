```python
a="1024170218"
L=[]
for i in range(0,10):
    L.append(10*int(a[i]))
print(L)
L.append(5)
print(L)
L.insert(0,5)
print(L)
L.pop()
print(L)
L.remove(0)
print(L)
L.sort()
L.sort(reverse=True)
print(L)
print("First 3 Elements:", L[:3])
print("Last 3 Elements:", L[-3:])
avg = sum(L) / len(L)
print("Average", avg)
```

```python
scores = tuple(L[:8])

print("Tuple scores", scores)
highest = max(scores)
lowest = min(scores)

print("Highest Score", highest)
print("Index", scores.index(highest))

print("Lowest Score", lowest)
print("Lowest Score", scores.count(lowest),)
reversed_list = list(scores[::-1])
print("Reversed Tuple", reversed_list)
num = int(input("Enter a score "))
if num in scores:
    print("First occurrence ", scores.index(num))
else:
    print("Score not present")
    first_score, second_score, *remaining_scores = scores

print("First Score:", first_score)
print("Second Score:", second_score)
print("Remaining Scores:", remaining_scores)
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
```

```python
import random

random.seed(1024170218)
numbers = [random.randint(100, 900) for _ in range(100)]

print("Random Numbers:")
print(numbers)

odd_numbers = [n for n in numbers if n % 2 != 0]
print("Odd Numbers:")
print(odd_numbers)
print("Count of Odd Numbers:", len(odd_numbers))

even_numbers = [n for n in numbers if n % 2 == 0]
print("\nEven Numbers:")
print(even_numbers)
print("Count of Even Numbers:", len(even_numbers))
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

prime_numbers = [n for n in numbers if is_prime(n)]

print("Prime Numbers:")
print(prime_numbers)
print("Count", len(prime_numbers))
most_frequent = max(set(numbers), key=numbers.count)

print("\nMost Frequent Number:", most_frequent)
print("Frequency:", numbers.count(most_frequent))
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 2]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)
print("\nUnion:", A.union(B))

print("Intersection:", A.intersection(B))

print( A.difference(B))
print( B.difference(A))
print("Symm Diff", A.symmetric_difference(B))

print( A.issubset(B))
print( B.issuperset(A))
x = int(input("\nEnter a value: "))
A.discard(x)

print("Updated Set A:", A)
```
