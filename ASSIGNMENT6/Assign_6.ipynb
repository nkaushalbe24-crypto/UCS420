```python

import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

fahrenheit = (9 / 5) * corrected_temperature + 32
print("Temperatures in Fahrenheit:", fahrenheit)

above_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", above_32)

count = np.sum(corrected_temperature > 32)
print("Number of readings greater than 32°C:", count)
```

```python

import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))

print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))

# (d) Total steps for each day
print("Total steps for each day:", np.sum(steps, axis=0))

# (e) Total steps for each user
print("Total steps for each user:", np.sum(steps, axis=1))

position = np.unravel_index(np.argmax(steps), steps.shape)
print("Position of maximum value:", position)
print("Maximum value:", steps[position])
```

```python

import numpy as np

# (a) Create original array
original = np.array([1, 2, 3, 4, 5, 6])

# (b) Slice elements from index 1 to 4
subset = original[1:4]
print("Subset:", subset)

# (c) Modify subset
subset[0] = 999
print("Original after modifying subset:", original)
print("Subset:", subset)

# (d) Create an independent copy
copied_subset = original[1:4].copy()
copied_subset[0] = 500

print("Original after modifying copy:", original)
print("Copied subset:", copied_subset)

# (e) Create a 3 x 4 matrix
matrix = np.arange(1, 13).reshape(3, 4)
print("3 x 4 matrix:\n", matrix)

# (f) Indexing and slicing
print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[1:3, 1:3])

# (g) Flatten using flatten() and ravel()
flat_array = matrix.flatten()
ravel_array = matrix.ravel()

print("Using flatten():", flat_array)
print("Using ravel():", ravel_array)

# (h) Modify ravel result
ravel_test = matrix.ravel()
ravel_test[0] = 1000
print("Matrix after modifying ravel result:\n", matrix)

# (i) Modify flatten result
flatten_test = matrix.flatten()
flatten_test[1] = 2000
print("Matrix after modifying flatten result:\n", matrix)
print("Flattened copy:", flatten_test)

# (j) Array properties
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

```python

import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)

y = np.array([40, 65, 30, 85], dtype=float)

# (a) Shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# (b) Transpose
XT = X.T
print("Transpose of X:\n", XT)

# (c) Matrix multiplication
XTX = XT @ X
print("X transpose multiplied by X:\n", XTX)

# (d) Inverse
XTX_inverse = np.linalg.inv(XTX)
print("Inverse of X.T @ X:\n", XTX_inverse)

# (e) Calculate OLS coefficients
beta = XTX_inverse @ XT @ y
print("OLS coefficients:", beta)

# (f) Interpretation
print("Coefficient 1 represents sleep hours.")
print("Coefficient 2 represents activity level.")
print("Coefficient 3 represents stress level.")

# (g) Predict score for a new user
new_user = np.array([5, 40, 7], dtype=float)
predicted_score = new_user @ beta

print("Predicted assistance score:", predicted_score)
```
